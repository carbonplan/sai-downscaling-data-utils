### Data storage

The datasets shared here live on S3 (via [source.coop](https://source.coop)) in the Icechunk format. [Icechunk](https://icechunk.io/en/stable/overview/) is an open-source format that adds `git`-like behavior to Zarr. Data in the Icechunk or Zarr formats aren't typical files - they're actually chunks of data that are stored in different nested directories. This means that you can't download them like you could a typical NetCDF file, for example. Instead, you interact with them by inspecting what is in each data "store" (the term used instead of "file") and accessing only the portion of the data you need. While very different from formats like NetCDF, it is tailored to working with large datasets in the cloud.

A single store holds every scenario and variable as a nested Zarr "group". The [glossary](https://github.com/carbonplan/sai-downscaling-data-utils/blob/main/GLOSSARY.md#storage-and-data-format) explains *store*, *group*, *chunk* and the other storage terms used here.

You don't need to understand the details. `load_downscaling_store(...)` loads the data into Xarray for you.

### How big a download request will be

Before pulling data, it helps to know what the store will actually read. Data is stored in "chunks", regularly-shaped arrays of data. You always have to read a the entire chunk containing your selected data, so a request touching even only a single cell or timepoint of a chunk still demands for the whole chunk.

| Property | Value |
|---|---|
| Shape (`ssp245`/`tas`) | `time=31046, lat=721, lon=1440`, float32, 129 GB |
| Native chunk | `(365, 36, 72)` = 3.8 MB |
| Chunk grid | 86 x 21 x 20 = 36,120 chunks |

One chunk holds **about a year of time over a 9° x 18° tile**. That shape decides which requests are cheap:

| Request | Logical | Chunks | Actually read |
|---|---|---|---|
| Delhi point, 85 yr | 0.12 MB | 86 | 0.33 GB |
| India box, 10 yr | 0.23 GB | 165 | 0.62 GB |
| Global, 10 yr, 1 var | 15.2 GB | 4,620 | 17.5 GB |
| Global, 10 yr, 2 vars | 30.3 GB | 9,240 | 35.0 GB |
| Global, single day | 4 MB | 420 | 1.6 GB |

Because time is chunked by the year, a **long time series at one point is comparatively expensive** — 85 years of the point at Delhi touches 86 chunks and reads 0.33 GB to deliver 0.12 MB of numbers. Conversely, a **map is cheap**: one global day reads 1.6 GB rather than the whole record.

> **Loading with `qa_flags=True` (see Section 5 below) roughly doubles every figure in the second table.** The `qa_flag_time_varying` array has the same shape and the same chunk grid as the data, so each selection reads it in full alongside. Section 5.5 covers when that is worth paying for.

> **These figures are for the downscaled product.** The bias-corrected product in Section 8 sits on a grid with about 19x fewer cells, so maps and global requests read far less — but a chunk there still spans a year, so a long single-point series barely gets cheaper.

Use `describe_request(...)` to check how big a selection is *before* you compute it. The cell below runs it on the four requests in the table. Read the absolute bytes rather than the ratio: a single global day amplifies its logical size roughly 400x but still only reads 1.6 GB, which is fine.

### Reduce before you load

The fix for an expensive request is almost never a bigger machine — it is asking for less. Any operations on the data (e.g. analysis, downloads) are supported by Dask, which operates "lazily", allowing most operations to happen smoothly with little RAM. A notable exception to this is operations that require the entire dataset in memory (e.g. distribution fitting)

| Step | Size |
|---|---|
| Global, 10 yr, daily | 15.2 GB |
| After `resample(time="YE").mean()` | 41.5 MB |
| After area-weighted `mean(["lat", "lon"])` | 40 bytes |

**Always weight spatial means by `cos(lat)`.** Grid cells are 0.25° on a side in *degrees*, but their true area shrinks toward the poles. An unweighted `.mean()` badly overweights high latitudes:

| Band | Share of grid cells | Share of true area |
|---|---|---|
| Poleward of 60° | 33.3% | 13.3% |
| Tropics (within 23.44°) | 25.9% | 39.7% |

Two details worth knowing:

- `.weighted()` renormalizes across NaN, so it stays correct over ocean-masked or clipped regions where a plain `.mean()` would not.
- `cos(lat)` is exact for a regular latitude grid. Polar cells get zero weight, which is harmless at 0.25°.

````python
# Reduce first, then compute, and weight spatial means by cos(lat)
# A region over a decade: ~0.56 GB read, so this cell does real work but stays modest.
region = ds[variable_name].sel(
    time=first_decade(ds), lat=slice(6, 38), lon=slice(68, 98)
)

# Reduce FIRST, then compute. Annual means collapse ~3,653 daily steps into 10.
annual = region.resample(time="YE").mean()

weights = np.cos(np.deg2rad(region.lat))
weighted_annual = annual.weighted(weights).mean(dim=["lat", "lon"])
unweighted_annual = annual.mean(dim=["lat", "lon"])

# Both reductions share one graph, so the region is read once, not twice.
reduced = xr.Dataset({"weighted": weighted_annual, "unweighted": unweighted_annual}).compute()
weighted_vals, unweighted_vals = reduced["weighted"], reduced["unweighted"]

print(f"weighted   annual mean: {float(weighted_vals.mean()):.2f} {region.attrs.get('units', '')}")
print(f"unweighted annual mean: {float(unweighted_vals.mean()):.2f} {region.attrs.get('units', '')}")
print(f"bias from forgetting weights: {float((unweighted_vals - weighted_vals).mean()):+.2f}")
print(f"  year-to-year spread of that bias: {float((unweighted_vals - weighted_vals).std()):.3f}")
print("\nA systematic offset, not noise -- which is why it must not be ignored.")
````

### 3.1 Point Lookup

**Use this method if**: You want time series data for specific locations (e.g., cities, weather stations, project sites).

Specify your target coordinates in the next cell. Longitudes run from -180 to 180, so a place at 100°W is `-100`.

`.sel(..., method="nearest")` picks the grid cell closest to your point, and it always returns one. If your point is outside the data, e.g. outside the region you subset to, you quietly get a cell at the edge instead, and a point outside a clipped outline gives you NaN. That's why each point lookup in this notebook runs `check_nearest(...)` first, with the same values. It stops with a message that says what went wrong.

````python
# ===== CUSTOMIZE: Set your target location =====
target_lat = 37.7749  # Latitude of your point of interest
target_lon = -122.4194  # Longitude of your point of interest, from -180 to 180
# ===============================================

# No modification needed below - extracts the point data
check_nearest(ds, lat=target_lat, lon=target_lon)  # stops with a clear message if the point is outside the data
ds_point_nearest = ds.sel(lat=target_lat, lon=target_lon, method="nearest")
print(f"Point subset shape: {dict(ds_point_nearest.sizes)}")
ds_point_nearest
````

````python
# ===== OPTIONAL: Extract multiple points at once =====
locations = {
    "San Francisco": (37.7749, -122.4194),
    "Los Angeles": (34.0522, -118.2437),
    "Seattle": (47.6062, -122.3321),
}

lats = xr.DataArray([lat for lat, lon in locations.values()], dims="location")
lons = xr.DataArray([lon for lat, lon in locations.values()], dims="location")
names = xr.DataArray(list(locations.keys()), dims="location")

check_nearest(ds, lat=lats, lon=lons)
multi_point = ds.sel(lat=lats, lon=lons, method="nearest").assign_coords(location=names)
print(multi_point)
````

### 3.3 Country/Region Mask

**Use this method if**: You want data clipped to country borders or administrative boundaries.

Specify your country/region name or provide a custom boundary file.

````python
# ===== CUSTOMIZE: Choose your boundary source =====
# Option A: local GIS / vector geometry file (recommended if you have one)
# Geopandas supports many vector format: geoparquet, shapefile, geopackage etc.
# from pathlib import Path
# boundaries_gpq = Path("path/to/your/boundaries.gpq")
# countries = gpd.read_parquet(boundaries_gpq)
# or
# countries = gpd.read_file(boundaries_shapefile)


# Option B: download Natural Earth (no local file needed)

countries_url = "https://naciscdn.org/naturalearth/50m/cultural/ne_50m_admin_0_countries.zip"
countries_path = pooch.retrieve(countries_url, known_hash=None, processor=pooch.Unzip())
countries = gpd.read_file([p for p in countries_path if p.endswith(".shp")][0])
# ==================================================

print(f"Available regions: {countries.shape[0]}")
countries[["NAME", "CONTINENT"]].head(10)
````

````python
# Attach a coordinate reference system, which rasterix needs
ds_with_crs = ds.proj.assign_crs(spatial_ref="epsg:4326")
ds_with_crs
````

````python
# ===== CUSTOMIZE: Select your country/region =====
country_name = "India"  # Change this to your country of interest
# ================================================

# No modification needed below - selects the country
country = countries[countries["NAME"] == country_name]
print(f"Selected country: {country_name}")
print(f"Bounds: {country.total_bounds}")  # [minx, miny, maxx, maxy]
country.plot(figsize=(8, 6), edgecolor="black", facecolor="lightblue")
````

#### Geometry Clip 

Clips to geometry bounds and masks values outside. Use this method for most workflows.

````python
# Clip to the country's outline


bounds = country.total_bounds
ds_bbox_country = (
    ds.sel(lat=slice(bounds[1], bounds[3]), lon=slice(bounds[0], bounds[2]))
    .sortby("lat", ascending=False)
    .unify_chunks()
)

ds_clipped = (
    geometry_clip(ds_bbox_country, country[["geometry"]], xdim="lon", ydim="lat", all_touched=True)
    .sortby("lat")
    .drop_vars("spatial_ref", errors="ignore")
)

# Safe to persist: the geometry clip has already cut this to one country.
describe_request(ds_clipped[variable_name], "country-clipped subset")
ds_clipped = ds_clipped.persist()
````

````python
# Plot one day of the clipped data
ds_clipped[variable_name].isel(time=100).squeeze().drop_vars("spatial_ref", errors="ignore").plot(
    figsize=(10, 6), robust=True, cmap="RdYlBu_r"
)
````

**Tips:**
- `.sel(..., method="nearest")` returns a cell even for a point outside your data. Run `check_nearest(ds, lat=..., lon=...)` first (Section 3.1)

- Run `describe_request(...)` (Section 2) on a selection before computing it — spatial subsetting is the cheapest lever you have on this store
- `geometry_clip` is recommended for most use cases as it reduces data volume
- Only `.persist()` after you have subset to a region that comfortably fits in memory

````python
# Standardize to z-scores; a template for your own transformations
def standardize(da, dim="time"):
    """Standardize data to z-scores along a dimension."""
    mean = da.mean(dim=dim)
    std = da.std(dim=dim)
    return (da - mean) / std


# Apply it
tas_zscore = standardize(ds_clipped[variable_name])
print(f"Z-score mean: {float(tas_zscore.mean()):.6f} (should be ~0)")
print(f"Z-score std: {float(tas_zscore.std()):.6f} (should be ~1)")
````

### 5.5 Quality flags

Each dataset contains three binary quality flags as ordinary data variables, alongside the data. `load_downscaling_store` drops them unless you pass `qa_flags=True`, and `qa_flag_vars(ds)` lists what a group carries.

| Flag | Shape | Contained in | What it marks |
|---|---|---|---|
| `qa_flag_time_varying` | `(time, lat, lon)` | every group except `dtr` | pixel-days failing a quality check: outlier screening, variable-specific plausible ranges, or physically impossible relationships such as `tasmax < tas` |
| `trend_distortion_flag` | `(lat, lon)` | every group except `dtr` | pixels where debiasing or spatial disaggregation meaningfully changed a scenario-intercomparison signal **relative to the raw GCM output**, evaluated in the ensemble mean |
| `qa_flag_time_invariant` | `(lat, lon)` | fewer than half of groups | pixels where debiasing or downscaling meaningfully changed a scenario-intercomparison signal, tested with a 5% threshold window |

All three are binary: `0` = no known issue, `1` = known issue.

The two time-invariant flags share the label "Change distortion flag" in the store, but they are **not interchangeable**. On `CESM2-WACCM6`/`bcsd`/`ssp245`/`tas`, member `003`, they flag 2.1% and 8.7% of pixels respectively. Check which one your analysis needs.

`trend_distortion_flag` does not share the `qa_flag_` prefix, so select flags with `qa_flag_vars(ds)` rather than by name.

**These are not free.** `qa_flag_time_varying` has the same shape *and the same chunk grid* as the data variable, so every selection that carries it reads a second array of the same dimensions — the size tables in Section 2 roughly double. It is `uint8` rather than `float32`, so it decompresses to a quarter of the bytes in memory, but the chunk count, and therefore the request count, is identical. The two time-invariant flags are single 2-D arrays and size next to nothing.

Load them when you intend to act on them, and drop them again before exporting if you do not want them in the file.

## 7. Global analysis on a small machine

Everything above works on a region, which is likely what most people need. If you need the whole globe, keep in mind that the amount of data a request reads can't be reduced: every chunk your selection touches is read in full, so a decade of one variable over the whole globe reads about 17.5 GB on any machine. What you can control is how much of it you hold in memory at once.

| Approach | Example | When to use it |
|---|---|---|
| **Load, then work in memory** | `ds[variable_name].sel(time=slice("2050-01-01", "2059-12-31"), lat=slice(6, 38), lon=slice(68, 98)).load()` | When your selection fits in memory. This India box over a decade is 0.23 GB. Check your own selection with `describe_request(...)` before loading it. |
| **Reduce, then compute** | `ds[variable_name].sel(time=slice("2050-01-01", "2059-12-31")).resample(time="YE").mean().compute()` | When your selection is too big to hold, which is usually the case for a global request. Only a few chunks are in memory at a time, and the result is 41.5 MB of annual means instead of 15.2 GB of daily values. |

For a global request, reduce first and then compute. Calling `.load()` on a whole global selection is the most common reason a notebook crashes.

For more detail, including how to run a global analysis on a laptop or an HPC system, what it measured and an example that makes a map, see [`compute-resources.ipynb`](compute-resources.ipynb).

## 8. Accessing coarse, bias-corrected data

Every store also contains the intermediate step *before* the spatial disaggregation step: the GCM output after bias correction, left on the GCM's coarse grid. Pass `product="debiased_coarse"` to any helper from Section 2 to use it — or set `product_name` there to run sections 3-7 on it.

**Use it when** you need to separate what bias correction did from what spatial disaggregation added, or when the GCM's native resolution is enough for your question.

| | `downscaled` | `debiased_coarse` |
|---|---|---|
| Group path | `{method}/{scenario}/{variable}/{member}` | `{method}/debiased_coarse/{scenario}/{variable}/{member}` |
| Grid, `CESM2-WACCM6` | 0.25° (721 × 1440) | 0.94° × 1.25° (192 × 288) |
| Grid, `UKESM1-1-LL` | 0.25° (721 × 1440) | 1.25° × 1.875° (144 × 192) |
| Chunk | `(365, 36, 72)`, 3.8 MB | `(365, 30, 60)`, 2.6 MB |
| Variables | `tas`, `tasmax`, `tasmin`, `pr`, `rsds` | the same, plus `dtr` |
| Members and coverage | as in Section 2 | identical |
| Quality flags | Section 5.5 | the same three; `dtr` has none |

Three things to know before using it:

- **It still depends on the method.** `bcsd` and `qdmsd` bias-correct differently — a nonparametric quantile mapping and quantile delta mapping respectively, as each group's `srm_downscaling:bias_correction_method` attribute records — so `method` matters here too.
- **It is cheaper for maps, not for long point series.** The grid has about 19x fewer cells, so wide requests read far less. A chunk still spans a year of time, though, so a single-point record touches the same number of chunks.
- **`dtr` is the diurnal temperature range, `tasmax − tasmin`,** in K like the two temperatures it relates (the store attaches no units to it). Section 8.2 explains why it exists only here.


### 8.1 What spatial disaggreagtion adds

A coarse cell is one number standing in for everything inside it. Comparing it with the 0.25° cells whose centres fall inside that same cell shows what spatial disaggregation contributes — and it depends heavily on the terrain.

The cell below takes a decade mean at two places: Delhi, on the flat Indo-Gangetic plain, and Kathmandu, in the Himalayan foothills.


````python
# ===== CUSTOMIZE: places to compare, ideally one flat and one mountainous =====
places = {
    "Delhi": (28.6, 77.2),  # Indo-Gangetic plain
    "Kathmandu": (27.72, 85.32),  # Himalayan foothills
}
# ==========================================================================

# No modification needed below
# dtr has no downscaled counterpart, so compare on tasmax if you picked it.
compare_var = "tasmax" if variable_name in COARSE_ONLY_VARIABLES else variable_name
ds_fine = load_downscaling_store(
    scenario_name, compare_var, gcm=gcm_name, method=method_name, member=member_name,
    product="downscaled",
)
ds_coarse = load_downscaling_store(
    scenario_name, compare_var, gcm=gcm_name, method=method_name, member=member_name,
    product="debiased_coarse",
)

decade = first_decade(ds_coarse)
dlat = float(abs(ds_coarse.lat[1] - ds_coarse.lat[0]))
dlon = float(abs(ds_coarse.lon[1] - ds_coarse.lon[0]))

lazy = {}
for name, (lat, lon) in places.items():
    check_nearest(ds_coarse, lat=lat, lon=lon)
    cell = ds_coarse[compare_var].sel(lat=lat, lon=lon, method="nearest")
    clat, clon = float(cell.lat), float(cell.lon)
    # Every 0.25 degree cell whose centre falls inside that one coarse cell.
    footprint = (
        ds_fine[compare_var]
        .sel(lat=slice(clat - dlat / 2, clat + dlat / 2), lon=slice(clon - dlon / 2, clon + dlon / 2))
        .sel(time=decade)
        .mean("time")
    )
    weights = np.cos(np.deg2rad(footprint.lat))
    lazy[name] = (
        cell.sel(time=decade).mean(),
        footprint.weighted(weights).mean(),
        footprint.min(),
        footprint.max(),
    )

# One graph, so each coarse cell and its footprint are read once.
(computed,) = dask.compute(lazy)

units = ds_fine[compare_var].attrs.get("units", "")
print(f"{compare_var}, {decade.start[:4]}-{decade.stop[:4]} mean in {units} "
      f"({gcm_name}/{method_name}/{scenario_name}, member {ensemble_member(scenario_name, compare_var, gcm_name, method_name, member_name, product="downscaled")})\n")
print(f"{'':11s} {'coarse':>8s}   {'0.25° cells inside that coarse cell':^36s}")
print(f"{'':11s} {'cell':>8s}   {'mean':>8s} {'min':>8s} {'max':>8s} {'spread':>8s}")
for name, (coarse_mean, fine_mean, fine_min, fine_max) in computed.items():
    print(f"{name:11s} {float(coarse_mean):8.2f}   {float(fine_mean):8.2f} {float(fine_min):8.2f} "
          f"{float(fine_max):8.2f} {float(fine_max - fine_min):8.2f}")
````

In flat terrain the two products tell the same story: the coarse value and the fine cells beneath it agree closely, and those fine cells barely differ from one another. In mountains they do not. With the places and settings above, the fine cells under Kathmandu's single coarse cell span roughly 24 K — valley floors and high ridges averaged into one number. That spread is the information spatial disaggregation adds, and it is an example of what a native-resolution analysis cannot see.

Note also that the coarse value and the mean of the fine cells are close rather than identical. Spatial disaggregation does not simply subdivide one coarse cell into smaller ones, so do not expect the fine field to average back exactly.


### 8.2 `dtr` and `tasmin`

`tasmin` is not bias-corrected directly. Correcting `tasmax` and `tasmin` independently can leave pairs where the minimum exceeds the maximum, so the pipeline corrects `tasmax` and the diurnal range `dtr` instead, then reconstructs `tasmin = tasmax − dtr`. That makes `dtr` an input to the product rather than a product in its own right.

It is available at native resolution only. After disaggregaton, the fine `tasmax`/`tasmin` pair gets a final reconciliation step that `dtr` never goes through, so a downscaled `dtr` would no longer equal their difference and would mislead anyone reading it as the diurnal range. **At 0.25°, compute `tasmax − tasmin` yourself.**

On the coarse grid the identity holds exactly, which the cell below checks. `dtr` is only in the members that have `tasmax`/`tasmin`, so the check uses the first member that has it.
